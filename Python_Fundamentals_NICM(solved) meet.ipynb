# 3. Operators

Operators are symbols used to perform operations on values.

Main types:
- Arithmetic Operators
- Comparison Operators
- Logical Operators
- Assignment Operators

### Arithmetic Operators

Arithmetic operators are used for mathematical calculations.

````python
a = 10
b = 3

print(a + b)   # Addition
print(a - b)   # Subtraction
print(a * b)   # Multiplication
print(a / b)   # Division
print(a // b)  # Floor Division
print(a % b)   # Modulus
print(a ** b)  # Power
````

### Comparison Operators

Comparison operators compare two values and return `True` or `False`.

````python
a = 10
b = 5

print(a > b)
print(a < b)
print(a == b)
print(a != b)
print(a >= b)
print(a <= b)
````

### Logical Operators

Logical operators are used to combine conditions.

- `and` → Both conditions must be True
- `or` → At least one condition must be True
- `not` → Reverses the result

````python
age = 25

print(age > 18 and age < 30)
print(age < 18 or age > 60)
print(not age > 18)
````

### Assignment Operators

Assignment operators are used to assign or update values.

````python
x = 10

x += 5
print(x)

x -= 3
print(x)

x *= 2
print(x)
````

# 6. Conditional Statements

Conditional statements are used to make decisions in a program.

Python provides:
- `if`
- `elif`
- `else`

### `if` Statement

`if` executes a block of code when a condition is `True`.

````python
age = 20

if age >= 18:
    print("You are eligible to vote")
````

### `if-else`

`else` executes when the `if` condition is `False`.

````python
age = 16

if age >= 18:
    print("Adult")
else:
    print("Minor")
````

### `if-elif-else`

Used when we have multiple conditions.

````python
marks = 75

if marks >= 90:
    print("Grade A+")
elif marks >= 75:
    print("Grade A")
elif marks >= 60:
    print("Grade B")
else:
    print("Grade C")
````

# 7. Loops

Loops are used to execute a block of code repeatedly.

Python has two main loops:
- `for` loop
- `while` loop

### `for` Loop

A `for` loop is used to iterate over a sequence or range of values.

````python
for i in range(1, 6):
    print(i)
````

### `range()`

`range()` generates a sequence of numbers.

`range(1, 6)` generates numbers from `1` to `5`.

````python
for i in range(1, 11):
    print(i)
````

### `while` Loop

A `while` loop runs as long as its condition is `True`.

````python
count = 1

while count <= 5:
    print(count)
    count += 1
````

### `break`

`break` is used to stop a loop.

````python
for i in range(1, 11):
    if i == 6:
        break
    print(i)
````

### `continue`

`continue` skips the current iteration and moves to the next iteration.

````python
for i in range(1, 6):
    if i == 3:
        continue
    print(i)
````

### Practice

````python
# Practice 1: Print numbers from 1 to 20
for i in range(1, 21):
    print(i)

# Practice 2: Print even numbers from 1 to 20
for i in range(2, 21, 2):
    print(i)

# Practice 3: Print the multiplication table of a number
number = int(input("Enter a number: "))
for i in range(1, 11):
    print(f"{number} x {i} = {number * i}")

# Practice 4: Find the sum of numbers from 1 to 10
total = sum(range(1, 11))
print("Sum =", total)

# Practice 5: Use break to stop a loop at 5
for i in range(1, 11):
    if i == 5:
        break
    print(i)
````

- Reuse code

Python functions can be broadly classified into:

1. **Built-in Functions** – Already provided by Python. (print, input, len, type etc.)

# 13. Comprehensions

Comprehensions provide a short and simple way to create collections from existing sequences.

Common types:
- List Comprehension
- Dictionary Comprehension
- Set Comprehension

### List Comprehension

Used to create a new list in a single line.

````python
numbers = [1, 2, 3, 4, 5]

squares = [x ** 2 for x in numbers]

print(squares)
````

### List Comprehension with Condition

We can also add a condition using `if`.

````python
numbers = [1, 2, 3, 4, 5, 6]

even_numbers = [x for x in numbers if x % 2 == 0]

print(even_numbers)
````

### Dictionary Comprehension

Used to create a dictionary in a single line.

````python
numbers = [1, 2, 3, 4, 5]

squares = {x: x ** 2 for x in numbers}

print(squares)
````

### Set Comprehension

Used to create a set in a single line.

````python
numbers = [1, 2, 2, 3, 3, 4]

squares = {x ** 2 for x in numbers}

print(squares)
````

### Practice

1. Create a list of squares from 1 to 10 using list comprehension.
2. Create a list of even numbers from 1 to 20.
3. Create a dictionary containing numbers from 1 to 5 and their squares.

````python
# Practice 1: List of squares from 1 to 10
squares = [x ** 2 for x in range(1, 11)]
print(squares)

# Practice 2: List of even numbers from 1 to 20
even_numbers = [x for x in range(1, 21) if x % 2 == 0]
print(even_numbers)

# Practice 3: Dictionary of numbers 1 to 5 and their squares
square_dict = {x: x ** 2 for x in range(1, 6)}
print(square_dict)
````

````python
# reversed()

numbers = [1, 2, 3, 4, 5]

print(list(reversed(numbers)))
````

````python
# all()

values = [True, True, True]

print(all(values))
````

````python
# any()

values = [False, False, True]

print(any(values))
````

````python
# isinstance()

x = 10

print(isinstance(x, int))
print(isinstance(x, str))
````

````python
# id()

x = 10

print(id(x))
````

````python
# dir()

text = "Python"

print(dir(text))
````
